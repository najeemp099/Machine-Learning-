## Handling Missing Values 

```python
df.dropna(subset=['Customer_ID'],inplace=True)
df.isnull().sum()
```

```python
df['Age'].unique()
```

```python
import re
def extract_age(age):
    age_num = re.findall(r'\d+',str(age))
    if len(age_num) > 0:
        return age_num[0]
    else :
        return age

df['Age'] = df['Age'].apply(lambda x : extract_age(x))
```

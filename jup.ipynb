```python
import matplotlib.pyplot as plt 
#car vs onroad price 
plt.figure(figsize=(14,8))
plt.bar(df["Name"],df["Onroad Price"],edgecolor = "black")
plt.title("Onroad Price of Cars")
plt.xlabel("Cars")
plt.ylabel("Price")
plt.xticks(rotation = 45,ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()




```

```python
#Mileage Comparison
plt.figure(figsize=(14,8))
plt.bar(df["Name"],df["Range (kmpl)"],edgecolor = "black")
plt.title("Mileage Comparison")
plt.xlabel("Car Name")
plt.ylabel("Mileage (kmpl)")
plt.xticks(rotation = 20,ha="right")
plt.grid(axis="y", linestyle = "--", alpha = 0.7)
plt.tight_layout()
plt.show()
```

```python
#Engine Capacity VS Mileage 
plt.scatter(df["CC"],df["Range (kmpl)"])
plt.title("Engine Capacity VS Mileage ")
plt.xlabel("Engine Capacity (CC)")
plt.ylabel("Range (kmpl)")
plt.show()
```

```python
#Exshowroom VS Onroad Price 
plt.figure(figsize=(10,7))
plt.plot(df["Name"],df["Ex-Showroom Price"],marker='o',label="Ex-showroom")
plt.plot(df["Name"],df["Onroad Price"],marker="*",label = "Onroad Price")

plt.title("Ex-showroom Vs Onroad Price")
plt.xlabel("Car")
plt.ylabel("Price ")
plt.xticks(rotation = 45)
plt.legend()
plt.tight_layout()
plt.show()

```

```python
#Distribution of Onroad Price 
plt.hist(df["Onroad Price"],bins =5,edgecolor="black")
plt.xlabel("Onroad Price ")
plt.ylabel("Frequency")
plt.show()
```
